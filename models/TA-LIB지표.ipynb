## 📈 1. 오실레이터 계열 (Momentum Indicators)
RSI (Relative Strength Index): 상대 강도 지수 — 추세의 강도를 측정.

MACD (Moving Average Convergence Divergence): 이동 평균 수렴·확산 — 추세 및 모멘텀 파악.

MACD_Signal: MACD의 시그널 라인.

MACD_Hist: MACD와 시그널 라인의 차이.

SlowK (Stochastic Oscillator %K): 현재 가격의 위치를 나타냄.

SlowD (Stochastic Oscillator %D): SlowK의 이동평균, 매수·매도 신호로 사용.

CCI (Commodity Channel Index): 과매수·과매도 상태 판단.

MFI (Money Flow Index): 거래량 기반 RSI로 자금 유입·유출을 측정.

ADX (Average Directional Index): 추세 강도를 측정.

## 📊 2. 이동 평균 계열 (Moving Averages)
SMA (Simple Moving Average): 단순 이동 평균 — 일정 기간 동안의 평균 가격.

EMA (Exponential Moving Average): 지수 이동 평균 — 최신 가격에 가중치 부여.

WMA (Weighted Moving Average): 가중 이동 평균 — 시간 가중치를 부여한 평균.

## 📈 3. 변동성 계열 (Volatility Indicators)
Bollinger Bands (BBANDS): 상한선(UpperBand), 중간선(MiddleBand), 하한선(LowerBand) — 변동성 분석.

SAR (Parabolic SAR): 추세 반전 시점을 포착.

ATR (Average True Range): 평균 변동 폭 — 변동성을 나타냄.

NATR (Normalized ATR): ATR을 정규화한 값.

TRANGE (True Range): 최대·최소 가격 차이 — 당일 변동폭.

## 📊 4. 거래량 계열 (Volume Indicators)
OBV (On-Balance Volume): 거래량 누적 — 거래량 흐름을 분석.

AD (Accumulation/Distribution Line): 매집·분배 — 가격과 거래량의 상관관계.

ADOSC (Chaikin A/D Oscillator): AD 라인의 속도 — 단기·장기 이동평균 차이.
