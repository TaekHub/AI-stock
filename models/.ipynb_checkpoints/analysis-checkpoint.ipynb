```python
import yfinance as yf
import pandas as pd

# 티커 설정 (예: PLTR)
ticker = "PLTR"

# 데이터 수집 (기간은 필요에 따라 조정)
df = yf.download(ticker, start="2020-01-01", end="2025-01-01", interval="1d")

# 기본 컬럼 선택
df = df[['Open', 'High', 'Low', 'Close', 'Volume']]

print("데이터 수집 완료:")
print(df.head())
```

```python
print(df)
```

```python
print(df.info())
```

```python
import numpy as np
import pandas as pd
import talib

# 1. Daily Return (전일 대비 종가 변화율)
daily_return = df['Close'].pct_change().values.ravel()  # 1D array

# 2. Range (고가 - 저가)
price_range = (df['High'] - df['Low']).values.ravel()

# 3. Volume Change (전일 대비 거래량 변화율)
volume_change = df['Volume'].pct_change().values.ravel()

# TA-Lib 함수는 1D numpy 배열을 요구하므로,
# 'Close'와 'Volume' 컬럼을 1D 배열로 변환 (float 타입, ravel()로 1D 보장)
close = np.array(df['Close'], dtype=float).ravel()
volume = np.array(df['Volume'], dtype=float).ravel()

# 4. OBV (On-Balance Volume)
obv = talib.OBV(close, volume).ravel()

# 5. RSI (Relative Strength Index, 기간 14)
rsi = talib.RSI(close, timeperiod=14).ravel()

# 6. MACD: MACD, MACD Signal, MACD Histogram
macd, macd_signal, macd_hist = talib.MACD(close, fastperiod=12, slowperiod=26, signalperiod=9)
# 각 결과를 1D 배열로 보장
macd = np.array(macd).ravel()
macd_signal = np.array(macd_signal).ravel()
macd_hist = np.array(macd_hist).ravel()

# 기술적 지표들을 딕셔너리 형태로 모아서 새로운 DataFrame 'ind' 생성 (df의 인덱스를 사용)
ind = pd.DataFrame({
    'daily_return': daily_return,
    'range': price_range,
    'volume_change': volume_change,
    'obv': obv,
    'rsi': rsi,
    'macd': macd,
    'macd_signal': macd_signal,
    'macd_hist': macd_hist
}, index=df.index)

# 초기 지표 계산으로 인한 NaN 값 제거
ind.dropna(inplace=True)

print(ind.head())
```

```python
# 데이터 타입을 numpy 배열로 변환
close = df['Close'].values  # pandas.Series를 numpy.ndarray로 변환
high = df['High'].values
low = df['Low'].values
volume = df['Volume'].values

# 각종 기술적 지표 계산
rsi = talib.RSI(close, timeperiod=14)  # close를 numpy.ndarray로 변환
macd, macd_signal, macd_hist = talib.MACD(close, fastperiod=12, slowperiod=26, signalperiod=9)
slowk, slowd = talib.STOCH(high, low, close, fastk_period=14, slowk_period=3, slowd_period=3)
cci = talib.CCI(high, low, close, timeperiod=14)
mfi = talib.MFI(high, low, close, volume, timeperiod=14)
adx = talib.ADX(high, low, close, timeperiod=14)
sma = talib.SMA(close, timeperiod=14)
ema = talib.EMA(close, timeperiod=14)
wma = talib.WMA(close, timeperiod=14)
upperband, middleband, lowerband = talib.BBANDS(close, timeperiod=20, nbdevup=2, nbdevdn=2, matype=0)
sar = talib.SAR(high, low, acceleration=0.02, maximum=0.2)
atr = talib.ATR(high, low, close, timeperiod=14)
natr = talib.NATR(high, low, close, timeperiod=14)
trange = talib.TRANGE(high, low, close)
obv = talib.OBV(close, volume)
ad = talib.AD(high, low, close, volume)
adosc = talib.ADOSC(high, low, close, volume, fastperiod=3, slowperiod=10)

# 지표들을 DataFrame에 추가
ind = pd.DataFrame({
    'RSI': rsi,
    'MACD': macd,
    'MACD_Signal': macd_signal,
    'MACD_Hist': macd_hist,
    '%K': slowk,
    '%D': slowd,
    'CCI': cci,
    'MFI': mfi,
    'ADX': adx,
    'SMA': sma,
    'EMA': ema,
    'WMA': wma,
    'UpperBand': upperband,
    'MiddleBand': middleband,
    'LowerBand': lowerband,
    'SAR': sar,
    'ATR': atr,
    'NATR': natr,
    'TRANGE': trange,
    'OBV': obv,
    'AD': ad,
    'ADOSC': adosc
}, index=df.index)

# 초기 지표 계산으로 인한 NaN 값 제거
ind.dropna(inplace=True)

# 시각화
plt.figure(figsize=(14, 16))

# RSI
plt.subplot(6, 3, 1)
plt.plot(ind['RSI'], label='RSI', color='blue')
plt.title('RSI')
plt.legend()

# MACD
plt.subplot(6, 3, 2)
plt.plot(ind['MACD'], label='MACD', color='blue')
plt.plot(ind['MACD_Signal'], label='MACD Signal', color='red')
plt.bar(ind.index, ind['MACD_Hist'], label='MACD Histogram', color='green', alpha=0.3)
plt.title('MACD')
plt.legend()

# Stochastic Oscillator
plt.subplot(6, 3, 3)
plt.plot(ind['%K'], label='%K', color='blue')
plt.plot(ind['%D'], label='%D', color='red')
plt.title('Stochastic Oscillator')
plt.legend()

# CCI
plt.subplot(6, 3, 4)
plt.plot(ind['CCI'], label='CCI', color='blue')
plt.title('CCI')
plt.legend()

# MFI
plt.subplot(6, 3, 5)
plt.plot(ind['MFI'], label='MFI', color='blue')
plt.title('MFI')
plt.legend()

# ADX
plt.subplot(6, 3, 6)
plt.plot(ind['ADX'], label='ADX', color='blue')
plt.title('ADX')
plt.legend()

# SMA
plt.subplot(6, 3, 7)
plt.plot(ind['SMA'], label='SMA', color='blue')
plt.title('SMA')
plt.legend()

# EMA
plt.subplot(6, 3, 8)
plt.plot(ind['EMA'], label='EMA', color='blue')
plt.title('EMA')
plt.legend()

# WMA
plt.subplot(6, 3, 9)
plt.plot(ind['WMA'], label='WMA', color='blue')
plt.title('WMA')
plt.legend()

# Bollinger Bands
plt.subplot(6, 3, 10)
plt.plot(ind['MiddleBand'], label='Middle Band', color='blue')
plt.plot(ind['UpperBand'], label='Upper Band', color='green')
plt.plot(ind['LowerBand'], label='Lower Band', color='red')
plt.title('Bollinger Bands')
plt.legend()

# Parabolic SAR
plt.subplot(6, 3, 11)
plt.plot(ind['SAR'], label='SAR', color='blue')
plt.title('Parabolic SAR')
plt.legend()

# ATR
plt.subplot(6, 3, 12)
plt.plot(ind['ATR'], label='ATR', color='blue')
plt.title('ATR')
plt.legend()

# NATR
plt.subplot(6, 3, 13)
plt.plot(ind['NATR'], label='NATR', color='blue')
plt.title('NATR')
plt.legend()

# TRANGE
plt.subplot(6, 3, 14)
plt.plot(ind['TRANGE'], label='TRANGE', color='blue')
plt.title('TRANGE')
plt.legend()

# OBV
plt.subplot(6, 3, 15)
plt.plot(ind['OBV'], label='OBV', color='blue')
plt.title('OBV')
plt.legend()

# AD
plt.subplot(6, 3, 16)
plt.plot(ind['AD'], label='AD', color='blue')
plt.title('AD')
plt.legend()

# ADOSC
plt.subplot(6, 3, 17)
plt.plot(ind['ADOSC'], label='ADOSC', color='blue')
plt.title('ADOSC')
plt.legend()

plt.tight_layout()
plt.show()
```
